```python
from ultralytics import YOLO
import cv2 as cv
import os
import pathlib
import glob
```

```python
model = YOLO("model/bottle_best_ver7.pt")
```

```python
dataset_path = pathlib.Path("datasets/bottle_detection/bottle")
test_result = pathlib.Path("test_result/bottle_detection/bottle")

img_paths = [str(path) for path in dataset_path.rglob("*") if path.is_file()]
print(img_paths)

for path in img_paths:
    img = cv.imread(path)
    result = model(img)
    annotated_img = result[0].plot()
    cv.imwrite(test_result.joinpath(pathlib.Path(path).name), annotated_img)


# img = [cv.imread(os.path.join(dataset_path,"0b1b1d6f-115.jpeg")), cv.imread(os.path.join(dataset_path,"0b1b1d6f-115.jpeg"))]
# img_paths = 



# results = model(img)
# annotated_img = results[0].plot()
# cv.imwrite(os.path.join(test_result, "test.jpg"), annotated_img)
```
